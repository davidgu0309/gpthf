```python
import torch

def create_boolean_tensor(ranges: torch.Tensor, max_size: int) -> torch.Tensor:
    assert ranges.dim() == 2 and ranges.size(1) == 2, "Input tensor must be of shape (B, 2)"
    
    B = ranges.size(0)
    result = torch.zeros((B, max_size), dtype=torch.bool)
    
    idx = torch.arange(max_size).expand(B, max_size)
    
    start, end = ranges[:, 0].unsqueeze(1), ranges[:, 1].unsqueeze(1)
    
    result = (idx >= start) & (idx < end)
    
    return result

def overwrite_with_mask(a: torch.Tensor, b: torch.Tensor, mask: torch.Tensor) -> torch.Tensor:
    # Flatten a and mask to use advanced indexing
    a_flat = a.view(-1)
    mask_flat = mask.view(-1)
    
    # Get the indices where mask is True
    indices = torch.nonzero(mask_flat, as_tuple=True)[0]
    
    # Overwrite values in a_flat at the positions of indices with values from b
    a_flat[indices] = b
    
    # Reshape a_flat back to the original shape of a
    a_overwritten = a_flat.view_as(a)
    
    return a_overwritten

# Example usage
ranges = torch.tensor([[1, 4], [2, 5], [0, 3]])
max_size = 6
mask = create_boolean_tensor(ranges, max_size)

# a is a tensor to be overwritten
a = torch.zeros((3, 6), dtype=torch.float32)

# b contains the values to overwrite a with, at positions specified by the mask
b = torch.tensor([10, 11, 12, 13, 14, 15, 16, 17, 18], dtype=torch.float32)

result = overwrite_with_mask(a, b, mask)
print(result)
```

```python
import torch

def generate_block_matrix(batch_vector):
    # Create a mask identifying non-padded entries
    non_padded = batch_vector != -1

    # Expand dimensions for broadcasting
    vec_expanded = batch_vector.unsqueeze(2)
    vec_transposed = batch_vector.unsqueeze(1)

    # Create a block matrix mask where each element attends to elements with the same value
    same_value_mask = (vec_expanded == vec_transposed)

    # Ensure that padded values (-1) only attend to other -1s
    padded_mask = (batch_vector == -1).unsqueeze(1) & (batch_vector == -1).unsqueeze(2)
    block_matrix_mask = same_value_mask & (non_padded.unsqueeze(1) & non_padded.unsqueeze(2)) | padded_mask

    # Detect changes including first element and non-padded transitions
    changes = torch.cat([torch.ones((batch_vector.shape[0], 1), dtype=torch.bool), batch_vector[:, 1:] != batch_vector[:, :-1]], dim=1)
    valid_changes = changes & non_padded

    # Determine the start indices for each sequence
    start_indices = valid_changes.nonzero(as_tuple=False)
    sequence_ids, counts = start_indices[:, 0], start_indices[:, 1]

    # Find max number of segments in any sequence
    unique_sequences, segment_counts = torch.unique(sequence_ids, return_counts=True)
    max_segments = segment_counts.max()

    # Prepare padded boundaries and boundary mask
    padded_boundaries = torch.full((batch_vector.shape[0], max_segments), -1, dtype=torch.long)
    boundary_mask = torch.zeros_like(padded_boundaries, dtype=torch.bool)

    # Fill padded boundaries and update boundary mask
    indices = (sequence_ids * max_segments + counts).tolist()  # Compute linear indices for filling
    padded_boundaries.view(-1)[indices] = start_indices[:, 1]
    boundary_mask.view(-1)[indices] = True

    return block_matrix_mask, padded_boundaries, boundary_mask

# Example usage:
batch_vector = torch.tensor([[0, 1, 2, -1, 2]])
block_matrix_mask, padded_boundaries, boundary_mask = generate_block_matrix(batch_vector)

print("Block Matrix Mask:\n", block_matrix_mask)
print("Padded Boundaries:\n", padded_boundaries)
print("Boundary Mask:\n", boundary_mask)
```
